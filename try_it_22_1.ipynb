```python
from tensorflow.keras.applications import ResNet50

model = ResNet50(weights="imagenet", include_top=False)
model.summary()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from google.colab import files

# Define a function to visualize the filters
def visualize_filters(layer_name, model, img_path):
    # Get the layer output
    layer_output = model.get_layer(layer_name).output

    # Create a model that maps the input image to the layer output
    intermediate_model = tf.keras.models.Model(inputs=model.input, outputs=layer_output)

    # Load and preprocess the example image
    img = tf.keras.utils.load_img(img_path, target_size=(224, 224))
    img_array = tf.keras.utils.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = tf.keras.applications.resnet50.preprocess_input(img_array)

    # Get the output of the intermediate model
    intermediate_output = intermediate_model.predict(img_array)

    # Visualize the feature maps (first 8 for simplicity)
    num_filters = intermediate_output.shape[-1]
    size = intermediate_output.shape[1]
    display_grid = np.zeros((size, size * min(num_filters, 8)))

    for i in range(min(num_filters, 8)):
        # Postprocess the filter to make it visually appealing
        filter_img = intermediate_output[0, :, :, i]
        filter_img = (filter_img - filter_img.mean()) / filter_img.std()
        filter_img = (filter_img * 64) + 128
        filter_img = np.clip(filter_img, 0, 255).astype('uint8')
        display_grid[:, i * size : (i + 1) * size] = filter_img

    plt.figure(figsize=(size, size / 8))
    plt.title(layer_name)
    plt.imshow(display_grid, aspect='auto', cmap='viridis')
    plt.axis('off')
    plt.show()

# --- Instructions for the user ---
# 1. Click on the folder icon on the left panel to open the file browser.
# 2. Click on the upload icon and select an image file (e.g., a JPG file).
# 3. Update the 'image_file_name' variable below with the name of your uploaded file.
# 4. Run this cell.

# Specify the name of the uploaded image file
image_file_name = 'my_image.jpg' # <--- Update this with your image file name

# Visualize the first convolution layer ('conv1_conv')
visualize_filters('conv1_conv', model, image_file_name)
```
